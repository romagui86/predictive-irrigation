# Fase 1: clima
Generamos lluvia, temperatura y ET0 Hargreaves. El pronóstico es distinto de la lluvia realizada.

```python
# Configurar imports para ejecutar desde la raíz o notebooks/.
from pathlib import Path
import sys
import matplotlib.pyplot as plt
ROOT = Path.cwd()
if not (ROOT / 'src').exists(): ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
from src.simulator import climate

# Generar una temporada reproducible con lluvia y pronósticos.
weather = climate(seed=7, regime='dry')
# Comparar lluvia, temperatura y demanda ET0 durante el ciclo.
fig, axes = plt.subplots(3, 1, figsize=(11, 7), sharex=True)
axes[0].bar(weather.t, weather.rain, label='Lluvia realizada')
axes[0].plot(weather.t, weather.rain_forecast, color='tab:orange', label='Pronóstico')
axes[1].plot(weather.t, weather.temp, label='Temperatura realizada')
axes[1].plot(weather.t, weather.temp_forecast, alpha=.7, label='Pronóstico')
axes[2].plot(weather.t, weather.et0, label='ET0 (mm/día)')
for ax in axes: ax.legend(); ax.grid(alpha=.2)
plt.tight_layout()
```
