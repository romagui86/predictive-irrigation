Se separa por sim_id para que una temporada no aparezca en más de un conjunto.

```python
# Preparar imports y ubicar la raíz del proyecto.
from pathlib import Path
import sys
import pandas as pd
ROOT = Path.cwd()
if not (ROOT / 'src').exists(): ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
from src.pipeline import split_ids, fit_models, predictive_metrics

# Leer tabla procesada y separar temporadas completas.
data = pd.read_csv(ROOT / 'data/processed/features_sinteticas.csv')
train_ids, val_ids, test_ids = split_ids()
train = data[data.sim_id.isin(train_ids)]
test = data[data.sim_id.isin(test_ids)]
# Entrenar en train y reportar métricas únicamente en test.
models = fit_models(train)
metrics, _ = predictive_metrics(test, models)
metrics
```
